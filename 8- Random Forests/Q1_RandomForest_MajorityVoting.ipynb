# Random Forest: Majority Voting Across Trees

**Question:** Given n decision trees built from different samples of a fruit basket, where
Tree-1 predicts Class-A (Apple), Tree-2 predicts Class-A (Apple), and Tree-n predicts Class-B
(Banana) — which fruit is likely to be picked as the final output?


```python
from collections import Counter
import matplotlib.pyplot as plt

# NAME: MADDIPATLA CHETAN
# ROLL NO:CH.SC.U4CSE24030

# Simulated votes from n individual decision trees (as shown in the figure)
tree_votes = ["Apple", "Apple", "Banana"]  # Tree-1, Tree-2, ..., Tree-n

vote_counts = Counter(tree_votes)
final_class = vote_counts.most_common(1)[0][0]

print("Votes per tree:", tree_votes)
print("Vote tally:", dict(vote_counts))
print("Final class after majority voting:", final_class)
```

```python
plt.bar(vote_counts.keys(), vote_counts.values(), color=["#e74c3c", "#f1c40f"])
plt.ylabel("Number of trees voting for this class")
plt.title("Random Forest Majority Voting")
plt.axhline(len(tree_votes) / 2, color="gray", linestyle="--", label="50% threshold")
plt.legend()
plt.show()
```

```python
# ---- General-purpose majority-vote function for any number of trees ----
def majority_vote(votes):
    counts = Counter(votes)
    winner, n_votes = counts.most_common(1)[0]
    return winner, n_votes, len(votes)

# Example with a larger simulated forest (n=7 trees)
larger_forest_votes = ["Apple", "Apple", "Banana", "Apple", "Banana", "Apple", "Apple"]
winner, n_votes, total = majority_vote(larger_forest_votes)
print(f"Final prediction: {winner} ({n_votes}/{total} trees agreed)")
```
